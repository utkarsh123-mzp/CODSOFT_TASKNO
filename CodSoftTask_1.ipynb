### Task 1: Data Cleaning & Preprocessing

```python
# 1. Import a dataset using Python and inspect its structure

df = pd.read_csv('netflix_movies (1).csv') 
print("--- Dataset Structure ---")
print(df.info())
print("\n--- First 5 Rows ---")
print(df.head())
```

```python
# 2. Identify missing values, duplicate records, and inconsistent data entries

print("\n--- Identifying Issues ---")
print("Missing values per column:\n", df.isnull().sum())
print(f"\nTotal duplicate rows: {df.duplicated().sum()}")
```

```python
# 3. Clean the dataset by handling null values, removing duplicates, and correcting data types

df_cleaned = df.drop_duplicates()
```

```python
# Basic cleaning strategy: Fill numeric nulls with mean, string nulls with 'Unknown'

for col in df_cleaned.columns:
    if df_cleaned[col].dtype in ['int64', 'float64']:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mean())
    else:
        df_cleaned[col] = df_cleaned[col].fillna('Unknown')
```

# 4. Prepare the data for further analysis using Pandas
# (The dataframe 'df_cleaned' is now fully preprocessed)

```python
# 5. Bonus: Save the cleaned dataset as a new CSV file

output_file = 'cleaned_netflix_dataset.csv'
df_cleaned.to_csv(output_file, index=False)
print(f"\nSuccess! Data cleaned and saved as '{output_file}'.")
```
