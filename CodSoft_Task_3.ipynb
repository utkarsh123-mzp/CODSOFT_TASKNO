```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

# Task 3: Data Visualization Dashboard

```python
# Load dataset
df = pd.read_csv('cleaned_netflix_dataset.csv')
```

```python
# Pre-processing: Extract year from 'date_added' for our line chart and scatter plot
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
df['year_added'] = df['date_added'].dt.year
```

```python
# Create a figure for the dashboard layout (3 rows, 2 columns)
fig = plt.figure(figsize=(18, 15))
fig.suptitle('Netflix Data Visualization Dashboard by Utkarsh Upadhyay', fontsize=22, fontweight='bold', y=0.98)
```

```python
# --- 1. Bar Chart: Top 10 Genres ---
plt.subplot(3, 2, 1)
top_genres = df['listed_in'].value_counts().head(10)
sns.barplot(y=top_genres.index, x=top_genres.values, palette='magma', hue=top_genres.index, legend=False)
plt.title('1. Bar Chart: Top 10 Content Genres', fontsize=14)
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
```

```python
# --- 2. Line Chart: Content Added Over The Years ---
plt.subplot(3, 2, 2)
content_trend = df['year_added'].value_counts().sort_index()
sns.lineplot(x=content_trend.index, y=content_trend.values, color='red', marker='o', linewidth=2)
plt.title('2. Line Chart: Trend of Content Added Over Years', fontsize=12)
plt.xlabel('Year Added')
plt.ylabel('Number of Titles Add')
plt.grid(True, linestyle='--', alpha=0.6)
```

```python
# --- 3. Pie Chart: Movies vs TV Shows ---
plt.subplot(3, 2, 3)
type_counts = df['type'].value_counts()
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'], explode=(0.05, 0))
plt.title('3. Pie Chart: Proportion of Movies vs TV Shows', fontsize=14)
plt.legend()
```

```python
# --- 4. Histogram: Distribution of Release Years[cite: 1] ---
plt.subplot(3, 2, 4)
sns.histplot(df['release_year'], bins=30, kde=True, color='teal')
plt.title('4. Histogram: Distribution of Release Years', fontsize=14)
plt.xlabel('Release Year')
plt.ylabel('Frequency')
plt.xlim(1990, 2024)
```

```python
# --- 5. Scatter Plot: Release Year vs Year Added[cite: 1] ---
plt.subplot(3, 2, 5)
sns.scatterplot(x='release_year', y='year_added', hue='type', data=df, palette='Set1', alpha=0.7)
plt.title('5. Scatter Plot: Release Year vs Year Added', fontsize=14)
plt.xlabel('Release Year')
plt.ylabel('Year Added')
plt.xlim(1990, 2024)
```

```python
# Adjust layout and show the dashboard
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
