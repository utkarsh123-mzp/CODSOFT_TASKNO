```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Author: Utkarsh Upadhyay
# Task 2: Exploratory Data Analysis (EDA)

```python
# --- 1. Load a dataset and examine its features using descriptive statistics ---

print("Loading dataset...")
df = pd.read_csv('cleaned_netflix_dataset.csv')
```

```python
print("\n--- Descriptive Statistics (Numerical Features) ---")
print(df.describe())
```

```python
print("\n--- Descriptive Statistics (Categorical Features) ---")
print(df.describe(include='object'))
```

```python
# --- 2. Identify trends, distributions, and relationships between variables ---
plt.figure(figsize=(8, 5))
sns.countplot(x='type', data=df, palette='pastel', hue='type', legend=False)
plt.title('Distribution of Content Types (Movies vs TV Shows)')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.histplot(df['release_year'], bins=30, kde=True, color='purple')
plt.title('Trend of Content Releases Over the Years')
plt.xlabel('Release Year')
plt.show()
```

```python
# --- 3. Detect outliers and unusual patterns within the data ---
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['release_year'], color='orange')
plt.title('Detecting Outliers in Release Years')
plt.show()
```

```python
# --- 4. Use summary statistics to answer key business questions ---
print("\n--- Business Question 1: Which top 5 countries produce the most content? ---")
top_countries = df['country'].value_counts().head(5)
print(top_countries)
```

```python
print("\n--- Business Question 2: What are the top 5 most common content ratings? ---")
top_ratings = df['rating'].value_counts().head(5)
print(top_ratings)
```

```python
# Visualizing a business question
plt.figure(figsize=(10, 5))
sns.barplot(x=top_countries.index, y=top_countries.values, palette='viridis', hue=top_countries.index, legend=False )
plt.title('Top 5 Content Producing Countries on Netflix')
plt.ylabel('Number of Titles')
plt.show()
```

# EDA Findings Report

### Content Distribution: An analysis of the content types reveals that Movies make up a significantly larger portion of the Netflix catalog compared to TV Shows.

### Release Trends & Outliers: The distribution of release years indicates a massive surge in content production over the last decade. Additionally, the boxplot successfully identified older, classic titles (mostly released before the year 2000) as statistical outliers in the dataset.

### Geographic Focus: The United States and India emerged as the top two content-producing countries, highlighting Netflix's primary regional markets and production investments.

### Target Audience: The most frequent content ratings on the platform are 'TV-MA' and 'TV-14', suggesting that the majority of Netflix's library is targeted toward mature audiences and teens rather than young children.
